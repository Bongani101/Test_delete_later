Type of duplicates identified

Exact duplicates - This line of code tests the exact row duplicates within the dataset where all column values are identical. More likely to occur during wide concatenation.

Key-based duplicates - This line of code tests key-based duplicates whereby the same CustomerId appears multiple times however with different attributes caused by data entry inconsistencies and data collection discrepancies.

```python
combined_df = combined_df.drop_duplicates()
```

```python
combined_df = combined_df.drop_duplicates(subset=["CustomerId"], keep="first")
```

Handling duplicates

Exact duplicates - This involves removing the exact duplicate leaving one copy, because exact duplicates do not provide new insights however they inflate the dataset size, potentially biasing statistical analysis and model training.

Handling Key-based duplicates - Keep the first valid occurence per CustomerId, this aids to avoid preserve maximum insights while enforcing uniqueness. Since the objective is customer based churn analysis, each customer should appear once in the dataset. Treating duplicated IDs as separate customers would violate entity integrity and distort downstream aggregation and modeling.

```python
print("Dataset size after duplicate testing:", combined_df.shape)
```

Approach justification - 

Duplicate analysis of the consolidated dataset identified two forms of duplication: exact row‑level duplicates and customer‑level duplication based on repeated CustomerId values. Exact duplication was detected where all attribute values were identical, while CustomerId‑based duplication reflected multiple records representing the same underlying customer entity. Quantification revealed two exact duplicates and eleven records with duplicated CustomerId values; the latter count includes the exact duplicates, as identical rows necessarily share the same identifier. Given that the analysis is conducted at the customer level, retaining multiple records per customer would bias aggregation and modelling. Accordingly, exact duplicates were removed first, followed by enforcement of CustomerId uniqueness. This reduced the dataset from 10,049 to 10,038 rows, confirming successful duplicate resolution and preservation of one‑to‑one correspondence between customers and records.

```python
#Identifies which variable contains missing values and the count of missing entries per column
combined_df.isna().sum()
```

```python
#Clear quantification of missing values
missing_values = combined_df.isna().sum()
missing_percent = (missing_values / len(combined_df)) *100

missing_summary = pd.DataFrame({"Missing Values": missing_values, "Missing Percentage (%)": missing_percent})

missing_summary[missing_summary["Missing Values"] > 0]
```

Impact of missing data - 

Missing values were identified mostly in numerical values such as Age, CreditScore, Balance and EstimatedSalary, moreover in categorical values such as Gender. Numerical values are important because they have a direct impact on churn behavior and will be utilized in aggregation and predictive modeling, these factors are significant analytically. Untreated missing values can result in incomplete feature vectors, skewed statistical summaries, and a smaller sample size through row deletion. If left untreated, missing values in the Gender variable could reduce the reliability of subsequent analysis by excluding affected records from modelling or aggregation processes. Additionally, missing gender information can distort group‑based summaries or churn comparisons by gender. 

```python
#Filling numerical values using median
numerical_columns = combined_df.select_dtypes(include=["int64", "float64"]).columns

for col in numerical_columns:
    if combined_df[col].isna().sum() > 0:
        combined_df[col] = combined_df[col].fillna(combined_df[col].median())
```

```python
#Filling categorical values with mode
if combined_df["Gender"].isna().sum() > 0:
    combined_df["Gender"] = combined_df["Gender"].fillna(combined_df["Gender"].mode()[0])
```

```python
#Confirming that all missing values have been treated
combined_df.isna().sum()
```

Justification of treating strategies

Missing values were detected and quantified across multiple numerical variables. Given their relevance to the end subsequent analysis, a median-based imputation strategy was applied to preserve dataset size and prevent bias. Moreover since Gender is a categorical variable, the missing value was therefore treated using mode imputation, which replaces the missing entry with the most frequently occuring category. This is suitable because the dominant class distribution, avoids unnecessary row deletion. After treatment it was confirmed that the dataset contains no remaining missing values, ensuring readiness for aggregation and modelling.

Task 6

6.1 Customer Segmentation (Age-Based)

```python
age_bins = [18, 30, 45, 60, combined_df["Age"].max() + 1]
age_labels = ["18-29", "30-44", "45-59", "60+"]

#Add age segment variable
combined_df["AgeSegment"] = pd.cut(
    combined_df["Age"],
    bins=age_bins,
    labels=age_labels,
    right=False
)

combined_df["AgeSegment"].value_counts()
```

Justification of Age Segmentation

The Age variable was transformed from a continuous numerical feature into a categorical segmentation variable (AgeSegment) using predefined interval bins. This allows for group based analysis while preserving underlysing age information. This structure supports increased interpretability of anlytical results and comparison of churn pattern across age groups. 

6.2 Value-Based Segmentation

```python
#Threshold for high-value customers (75th percentile)
high_value_threshold = combined_df["Balance"].quantile(0.75)

#Add value-based customer segment
combined_df["ValueSegment"] = pd.cut(
    combined_df["Balance"],
    bins=[-1, 0, high_value_threshold, combined_df["Balance"].max()],
    labels=["Low Value", "Medium Value", "High Value"]
)


combined_df["ValueSegment"].value_counts()
```

Justification of Value-Based Segmentation

For the customer value the Balance variable was used, which was then transformed into ordinal categorical variable (ValueSegment). Due to thier low financial engagement , customers with zero balances were specifically categorised as Low Value. High Value clients with above average balances were identified by segmenting the remaining customers using a quantile-based threshold; all other customers were classified as Medium Value. In order to facilitate straightforward comparison, aggregation, and analysis of customer behavior across value tiers, this transformation generates exhaustive and mutually exclusive groups.

Task 7

```python
#Identifes how customer value differs across age groups
avg_balance_by_age = (
    combined_df
    .groupby("AgeSegment")["Balance"]
    .mean().reset_index(name="Average Balance")
)

avg_balance_by_age
```

```python
#Evaluates whether Credit Score differs between churned and retained customers
mean_credit_by_churn = (
    combined_df
    .groupby("Churn")["CreditScore"].
    mean().reset_index(name="Mean Credit Score")
)

mean_credit_by_churn
```

```python
#Compares churn behaviour across geographic regions while outline gender per region
churn_rate_geo_gender = (
    combined_df.groupby(["Geography", "Gender"])["Churn"]
    .agg(
        Churn_Rate="mean",
        Customer_Count="count"
    )
    .reset_index()
)

churn_rate_geo_gender
```

```python
#Outlines the number of customers by value
customer_count_by_value = (
    combined_df
    .groupby("ValueSegment")
    .size()
    .reset_index(name="Customer_Count")
)

customer_count_by_value
```

```python
#Outlines how balance differs across customer age and value tiers
avg_balance_age_value = (
    combined_df
    .groupby(["AgeSegment", "ValueSegment"])["Balance"]
    .mean()
    .reset_index(name="Average Balance")
)

avg_balance_age_value
```
